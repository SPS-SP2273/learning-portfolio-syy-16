### Task 4

```python
def rename_my_file(old_file_name):

    months = {
    'jan': '01', 'feb': '02', 'mar': '03',
    'apr': '04', 'may': '05', 'jun': '06',
    'jul': '07', 'aug': '08', 'sep': '09',
    'oct': '10', 'nov': '11', 'dec': '12'
    }

    my_files = ['oct-08_singapore.txt', 'jul-10_cairo.txt', 'may-15_dhaka.txt',
     'may-13_cairo.txt', 'oct-21_cairo.txt', 'jan-10_singapore.txt',
    'jun-20_tokyo.txt', 'aug-06_jakarta.txt', 'dec-21_karachi.txt',
    'jan-01_tokyo.txt']

    if old_file_name in my_files:
        return old_file_name

    for a,b in months.items():
        if a in old_file_name.lower():
            
            if old_file_name not in my_files:
                new_file_name = old_file_name.replace(a,b)
                final_filename = new_file_name.replace('_', '-')
                return final_filename
        
    if a not in old_file_name.lower():
        return 'no month abbreviation found'
```

```python
files = glob(os.path.join('os-collaboration-exercise-data', '*-*_*.txt'))

for f in files:
    txt_files = os.path.split(f)[-1]
    new_txt_file = rename_my_file(txt_files)
    
    if new_txt_file == txt_files:
        print(f'Skip{txt_files} (renamed)')
    if new_txt_file == 'no month abbreviation found':
        print(f'Skip{txt_files} (no months abbreviation found)')
    
    new_path = os.path.join('os-collaboration-exercise-data', new_txt_file)
    os.rename(f, new_path)
    print(f'{txt_files} --> {new_txt_file}')
```

### Task 6

```python
cairo_files = glob(os.path.join('os-collaboration-exercise-data', '*cairo.*'))

for c in cairo_files:
    file_to_copy = os.path.split(c)[-1]
    destination = os.path.join('os-collaboration-exercise-data', 'Cairo', file_to_copy)
    shutil.copy(c, destination)
    print(f'{c} --> {destination}')
```

### Task 7

```python
cities = ['Dhaka', 'Jakarta', 'Karachi', 'Manila',
 'Melbourne', 'Osaka', 'Shanghai', 'Singapore', 'Tokyo']  

for c in cities:
  file_to_transfer = glob(os.path.join('os-collaboration-exercise-data', f'*{c}*'))
  for t in file_to_transfer:
    final_destination = os.path.join('os-collaboration-exercise-data', c)
    shutil.move(t,final_destination)
    print(f'{t} --> {final_destination}')
```
